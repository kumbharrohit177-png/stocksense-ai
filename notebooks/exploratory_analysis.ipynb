**Lab Manual Focus:** Historical Data Ingestion, Stationarity Analysis (ADF), and Technical Indicators.

```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from backend.ml.preprocessing.data_cleaner import DataCleaner
from backend.ml.preprocessing.stationarity import StationarityAnalyzer
from backend.ml.feature_engineering.technical_indicators import FeatureEngineer

# 1. Download Real Historical Stock Data
symbol = 'RELIANCE.NS'
df = yf.download(symbol, period='2y', interval='1d')
clean_df = DataCleaner.clean_ohlcv(df)
print('Loaded Clean Dataset Shape:', clean_df.shape)
clean_df.head()
```

```python
# 2. Stationarity Analysis (Augmented Dickey-Fuller Test)
adf_raw = StationarityAnalyzer.adf_test(clean_df['Close'])
print('ADF Test on Raw Close Price:', adf_raw)

diff_series = clean_df['Close'].diff().dropna()
adf_diff = StationarityAnalyzer.adf_test(diff_series)
print('ADF Test on 1st Differenced Series:', adf_diff)
```

```python
# 3. Compute Engineered Technical Indicators
feat_df = FeatureEngineer.compute_features(clean_df).dropna()
print('Engineered Features Matrix Shape:', feat_df.shape)
print('Available Feature Columns:', list(feat_df.columns))
feat_df[['Close', 'MA_7', 'MA_21', 'MA_50', 'RSI_14', 'MACD', 'Volatility_21']].tail()
```
