# StockSense AI - Linear Regression Model
**Lab Manual Focus:** Feature Scaling, Chronological Split (80/20), Fitting OLS/Ridge, and Test Metric Evaluation (MAE, RMSE, R²).

```python
import yfinance as yf
import pandas as pd
from backend.ml.linear_regression.linear_model import LinearRegressionPredictor
from backend.app.services.data_service import DataService

# Load historical data
symbol = 'RELIANCE.NS'
df = DataService.fetch_historical_dataframe(symbol, period='2y')

# Train & Evaluate Model
lr_model = LinearRegressionPredictor(alpha=1.0)
eval_result = lr_model.train_and_evaluate(df, train_ratio=0.8)

print('=== LINEAR REGRESSION TEST METRICS ===')
for k, v in eval_result['metrics'].items():
    print(f'{k}: {v}')

# Generate 7-day Future Forecast
forecast = lr_model.predict_future(df, horizon_days=7)
print('\nPredicted Target Price (T+7):', forecast['predicted_price'])
print('Predicted Movement:', forecast['predicted_change_percent'], '% (', forecast['trend'], ')')
pd.DataFrame(forecast['forecast'])
```
