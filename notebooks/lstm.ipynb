# StockSense AI - Long Short-Term Memory (LSTM) Deep Learning Model
**Lab Manual Focus:** Normalization, 30-day Sliding Lookback Window, Recurrent Gating, Inverse Transformation, and Future Multi-Step Projection.

```python
import pandas as pd
from backend.ml.lstm.lstm_model import LSTMPredictor
from backend.app.services.data_service import DataService

# Load Data
symbol = 'RELIANCE.NS'
df = DataService.fetch_historical_dataframe(symbol, period='2y')

# Initialize & Train LSTM Network
lstm_model = LSTMPredictor(lookback_window=30, hidden_units=32, epochs=25, learning_rate=0.01)
eval_result = lstm_model.train_and_evaluate(df, train_ratio=0.8)

print('=== LSTM TEST EVALUATION METRICS ===')
for k, v in eval_result['metrics'].items():
    print(f'{k}: {v}')

# Generate Recursive Trajectory
forecast = lstm_model.predict_future(df, horizon_days=7)
print('\nLSTM Predicted 7-Day Target:', forecast['predicted_price'])
print('Predicted Trend Direction:', forecast['trend'])
pd.DataFrame(forecast['forecast'])
```
